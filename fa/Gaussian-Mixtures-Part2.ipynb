## الگوریتم K-Means: مرور

تابع امتیاز *مجموع مربع خطا* چنین تعریف می‌شود

$$
SSE(\mathcal{C}) = \sum_{i=1}^{k}\sum_{\mathbf{x}_{j} \in
C_{i}}\|\mathbf{x}_{j}-\mathbf{\mu}_i\|^2
$$

هدف یافتن خوشه‌بندی‌ای است که امتیاز SSE را کمینه کند:

$$
\mathcal{C}^* = \arg \min_{\mathcal{C}} \{ SSE(\mathcal{C}) \}
$$

K-means رویکردی تکراری و حریصانه برای یافتن خوشه‌بندی‌ای دارد که تابع هدف SSE را کمینه کند. از این‌رو
می‌تواند به بهینهٔ محلی برسد، نه خوشه‌بندی بهینهٔ کلی.

---

## الگوریتم K-Means: گام‌ها

K-means میانگین خوشه‌ها را با تولید تصادفی $k$ نقطه در فضای داده مقداردهی می‌کند. هر تکرار دو گام دارد:

1. **تخصیص خوشه:** هر نقطهٔ $\mathbf{x}_{j} \in \mathbf{D}$ به نزدیک‌ترین میانگین تخصیص می‌یابد و
   خوشه‌بندی‌ای می‌سازد که در آن هر خوشهٔ $C_i$ نقطه‌های نزدیک‌تر به $\mathbf{\mu}_i$ را دارد:

   $$ 
   i^* = \arg \min_{i=1}^k \Bigl\{\|\mathbf{x}_{j} - \mathbf{\mu}_i\|^2\Bigr\}
   $$

2. **به‌روزرسانی مرکز:** مقدارهای میانگین تازه برای هر خوشه از نقطه‌های $C_i$ حساب می‌شود.

این گام‌ها تا همگرایی به نقطهٔ ثابت یا کمینهٔ محلی تکرار می‌شوند.

## خوشه‌بندی با امید-بیشینه‌سازی

$X_a$ متغیر تصادفی متناظر با صفت $a$ام باشد. $\mathbf{X} = (X_1, X_2, \dots, X_d)$ متغیر تصادفی برداری
روی $d$ صفت باشد و $\mathbf{x}_{j}$ نمونه‌ای داده از $\mathbf{X}$.

فرض می‌کنیم هر خوشهٔ $C_i$ با توزیعی نرمال چندمتغیره مشخص می‌شود:

$$
f_i(\mathbf{x}) = f(\mathbf{x}|\mathbf{\mu}_i,\mathbf{\Sigma}_i)
= \frac{1}{(2\pi)^\frac{d}{2}|\mathbf{\Sigma}_i|^\frac{1}{2}}
    \exp\left\{-\frac{(\mathbf{x}-\mathbf{\mu}_{i})^{T} \mathbf{\Sigma}_i^{-1}
    (\mathbf{x}-\mathbf{\mu}_{i})}{2}\right\}
$$

که $\mathbf{\mu}_i \in \mathbb{R}^d$ میانگین خوشه و $\mathbf{\Sigma}_i \in \mathbb{R}^{d \times d}$
ماتریس کوواریانس است.


تابع چگالی احتمال $\mathbf{X}$ به‌صورت یک *مدل مخلوط گاوسی* روی همهٔ $k$ خوشه داده می‌شود:

$$
f(\mathbf{x}) = \sum_{i=1}^k f_i(\mathbf{x}) P(C_i)
= \sum_{i=1}^k f(\mathbf{x} \mid \mathbf{\mu}_i, \mathbf{\Sigma}_i) P(C_i)
$$

که احتمال‌های پیشین $P(C_i)$ را *پارامترهای مخلوط* می‌نامند و باید شرط زیر را برآورده کنند:

$$
\sum_{i=1}^k P(C_i) = 1
$$

### برآورد بیشترین درست‌نمایی

مجموعهٔ همهٔ پارامترهای مدل را به‌صورت فشرده چنین می‌نویسیم:

$$
\mathbf{\theta} = \left\{\mathbf{\mu}_1,\mathbf{\Sigma}_1,P(C_1), \dots, \mathbf{\mu}_k, \mathbf{\Sigma}_k, P(C_k)\right\}
$$

با داشتن مجموعه‌دادهٔ $\mathbf{D}$، *درست‌نمایی* $\mathbf{\theta}$ چنین است:

$$
P(\mathbf{D} | \mathbf{\theta}) = \prod_{j=1}^n f(\mathbf{x}_{j})
$$

[MLE](MLE-Introduction.md) پارامترهای $\mathbf{\theta}$ را چنان برمی‌گزیند که درست‌نمایی را بیشینه کنند:

$$
\mathbf{\theta}^* = \arg\max_{\mathbf{\theta}} \{\ln P(\mathbf{D} | \mathbf{\theta})\}
$$

که تابع لگاریتم درست‌نمایی چنین است:

$$
\ln P(\mathbf{D} | \mathbf{\theta}) = \sum_{j=1}^n \ln f(\mathbf{x}_{j})
    = \sum_{j=1}^n \ln \biggl( \sum_{i=1}^k f(\mathbf{x}_{j} | \mathbf{\mu}_i, \mathbf{\Sigma}_i)P(C_i)\biggr)
$$

---

### بیشینه کردن لگاریتم درست‌نمایی با امید-بیشینه‌سازی (EM)

بیشینه کردن مستقیم لگاریتم درست‌نمایی روی $\mathbf{\theta}$ دشوار است چون **جمع درون لگاریتم** هست:

$$
\ln P(\mathbf{D} | \mathbf{\theta}) = \sum_{j=1}^n \ln \left( \sum_{i=1}^k f(\mathbf{x}_{j} | \mathbf{\mu}_i, \mathbf{\Sigma}_i) P(C_i) \right)
$$

اگر تابع درست‌نمایی فقط یک **گاوسی تنها** داشت، مشتق‌گیری مستقیم سرراست بود. ولی در مدل مخلوط گاوسی،
حضور چند گاوسی درون جمع مشتق‌گیری را پیچیده می‌کند. چون هر نقطهٔ دادهٔ $\mathbf{x}_j$ به‌طور قطع به یک
خوشه تعلق ندارد و به‌جای آن احتمالی برای تعلق به چند خوشه دارد، حل تحلیلی پارامترها دشوار است.

#### EM چگونه کمک می‌کند؟

امید-بیشینه‌سازی رویکردی ساخت‌یافته برای بیشینه کردن *غیرمستقیم* درست‌نمایی می‌دهد، با وارد کردن
متغیرهای پنهان — تخصیص‌های خوشه $w_{ij} = P(C_i | \mathbf{x}_j)$. **گام امید** این احتمال‌ها را حساب
می‌کند و عملاً عضویت‌های پنهان را برآورد می‌کند، و **گام بیشینه‌سازی** پارامترها را با این وزن‌ها به‌روز
می‌کند.

با تکرار میان این گام‌ها، EM از پیچیدگی مشتق‌گیری مستقیم عبارت لگاریتمِ جمعِ تودرتو دوری می‌کند و برآورد
پارامترها را شدنی می‌سازد.

## امید-بیشینه‌سازی برای مدل‌های مخلوط گاوسی

### حالت یک‌بعدی: خوشه‌بندی نرمال تک‌متغیره

$\mathbf{D}$ شامل یک صفت $X$ باشد که هر نقطهٔ $x_{j} \in \mathbb{R}$ نمونه‌ای تصادفی از $X$ است. مدل
مخلوط برای هر خوشه از توزیع‌های نرمال تک‌متغیره استفاده می‌کند:

$$
f_i(x) = f(x | \mu_i, \sigma_i^2) =
  \frac{1}{\sqrt{2\pi}\sigma_i} \exp \left\{-\frac{(x -
  \mu_i)^2}{2\sigma_i^2} \right\}
$$

که پارامترهای خوشه شامل $\mu_{i}$ (میانگین)، $\sigma_{i}^2$ (واریانس) و $P(C_i)$ (احتمال پیشین) است.

#### مقداردهی اولیه
هر خوشهٔ $C_i$ (برای $i = 1, 2, \dots, k$) به‌طور تصادفی با پارامترهای $\mu_{i}$، $\sigma_{i}^2$ و
$P(C_i)$ مقداردهی می‌شود.

#### گام امید
با میانگین $\mu_i$، واریانس $\sigma_i^2$ و احتمال پیشین $P(C_i)$ هر خوشه، **احتمال پسین خوشه** با
قضیهٔ بیز حساب می‌شود:

$$
w_{ij} = P(C_{i}|x_{j}) =
    \frac{f(x_{j}|\mu_i,\sigma_i^2) \cdot P(C_{i})}
    {\sum_{a=1}^k f(x_{j} |\mu_a, \sigma_a^2) \cdot P(C_{a})}
$$

#### گام بیشینه‌سازی
با $ w_{ij} $، پارامترهای به‌روزشدهٔ خوشه چنین حساب می‌شوند:

- **به‌روزرسانی میانگین** (میانگین وزنی نقطه‌ها):

  $$
  \mu_{i} = \frac{\sum_{j=1}^n w_{ij} \cdot x_{j}}
    {\sum_{j=1}^n w_{ij}}
  $$

- **به‌روزرسانی واریانس** (واریانس وزنی نقطه‌ها):

  $$
  \sigma_{i}^{2} =
    \frac{\sum_{j=1}^n w_{ij}(x_{j}-\mu_{i})^2}
        {\sum_{j=1}^n w_{ij}}
  $$

- **به‌روزرسانی احتمال پیشین** (کسر وزن‌هایی که به خوشه سهم می‌دهند):

  $$
  P(C_{i}) =
    \frac{\sum_{j=1}^n w_{ij}}{n}
  $$

### حالت چندمتغیره: مدل مخلوط گاوسی با کوواریانس کامل

در **حالت چندمتغیره**، هر خوشهٔ $C_i$ با توزیعی نرمال چندمتغیره مدل می‌شود و از حالت یک‌بعدی فراتر
می‌رود:

$$
f_i(\mathbf{x}) = f(\mathbf{x}|\mathbf{\mu}_i,\mathbf{\Sigma}_i)
= \frac{1}{(2\pi)^\frac{d}{2}|\mathbf{\Sigma}_i|^\frac{1}{2}}
    \exp\left\{-\frac{(\mathbf{x}-\mathbf{\mu}_{i})^{T} \mathbf{\Sigma}_i^{-1}
    (\mathbf{x}-\mathbf{\mu}_{i})}{2}\right\}
$$

که $\mathbf{\mu}_i \in \mathbb{R}^d$ میانگین خوشه و $\mathbf{\Sigma}_i \in \mathbb{R}^{d \times d}$
ماتریس کوواریانس است.

#### گام امید (چندمتغیره)
مانند حالت تک‌متغیره، احتمال پسین خوشه با قضیهٔ بیز حساب می‌شود:

$$
w_{ij} = P(C_i | \mathbf{x}_{j}) = \frac{f_i(\mathbf{x}_{j}) \cdot P(C_{i})}
    {\sum_{a=1}^k f_a(\mathbf{x}_{j}) \cdot P(C_{a})}
$$

#### گام بیشینه‌سازی (چندمتغیره)

- **به‌روزرسانی میانگین** (میانگین وزنی نقطه‌ها):

  $$
  \mathbf{\mu}_{i} = \frac{\sum_{j=1}^n P(C_i | \mathbf{x}_j) \cdot \mathbf{x}_{j}}
  {\sum_{j=1}^n P(C_i | \mathbf{x}_j)}
  $$

- **به‌روزرسانی کوواریانس** (ماتریس کوواریانس وزنی روی ابعاد):

  $$
  \mathbf{\Sigma}_{i} =
    \frac{\sum_{j=1}^n P(C_i | \mathbf{x}_j) (\mathbf{x}_j - \mathbf{\mu}_i)(\mathbf{x}_j - \mathbf{\mu}_i)^T}
    {\sum_{j=1}^n P(C_i | \mathbf{x}_j)}
  $$

- **به‌روزرسانی احتمال پیشین** (کسر سهم کل در خوشه):

  $$
  P(C_{i}) =
    \frac{\sum_{j=1}^n P(C_i | \mathbf{x}_j)}{n}
  $$

## K-Means به‌عنوان حالت خاصی از امید-بیشینه‌سازی (EM)

K-means را می‌توان **نسخهٔ ساده‌شدهٔ** الگوریتم امید-بیشینه‌سازی (EM) دانست که در آن تخصیص‌های خوشه
قطعی‌اند نه احتمالاتی. به‌ویژه در K-means هر نقطهٔ دادهٔ $\mathbf{x}_{j}$ بر پایهٔ کمترین مربع فاصله تا
مرکز خوشه، انحصاراً به یک خوشه تعلق دارد:

$$
P(\mathbf{x}_{j} | C_i) =
\begin{cases}
  1 & \text{if } C_i = \arg\min_{C_a} \left\{
    \|\mathbf{x}_{j} - \mathbf{\mu}_a\|^2\right\}\\
  0 & \text{otherwise}
\end{cases}
$$

احتمال پسین تخصیص خوشه با قضیهٔ بیز حساب می‌شود:

$$
P(C_i | \mathbf{x}_{j}) = \frac{P(\mathbf{x}_{j} | C_i) P(C_i)}
{\sum_{a=1}^k P(\mathbf{x}_{j} | C_a) P(C_a)}
$$

چون K-means هر نقطه را قطعی به دقیقاً یک خوشه تخصیص می‌دهد، این به حالت‌های زیر ساده می‌شود:

- اگر $P(\mathbf{x}_{j} | C_i) = 0$، آن‌گاه $P(C_i | \mathbf{x}_{j}) = 0$.
- اگر $P(\mathbf{x}_{j} | C_i) = 1$، آن‌گاه برای همهٔ $a \neq i$ داریم
  $P(\mathbf{x}_{j} | C_a) = 0$، پس:

  $$
  P(C_i | \mathbf{x}_{j}) = \frac{1 \cdot P(C_i)}{1 \cdot P(C_i)} = 1.
  $$

پس در K-means عضویت خوشه دودویی است:

$$
P(C_i | \mathbf{x}_{j}) =
\begin{cases}
  1 & \text{if } \mathbf{x}_{j} \in C_i, \text{i.e., if }
  C_i = \arg\min_{C_a} \left\{ \|\mathbf{x}_{j} - \mathbf{\mu}_a\|^2\right\}\\
  0 & \text{otherwise}
\end{cases}
$$

برخلاف EM که ماتریس‌های کوواریانس را برای مدل کردن توزیع خوشه‌ها به کار می‌برد، K-means **فقط مرکزهای
خوشه $\mathbf{\mu}_i$ و نسبت‌های خوشه $P(C_i)$ را برآورد می‌کند**.
